Boolean Function: $ f(X_1, X_2, X_3) = X_1.X_2.X_3 + \overline{X_3}.X_4 + \overline{X_2}.X_4 $

The following cells display the BDDs for variable orderings of the Boolean function

1. $ X_2 < X_3 < X_4 < X_1 $
2. $ X_1 < X_2 < X_3 < X_4 $
3. $ X_2 < X_3 < X_1 < X_4 $

```python
f = boolfunc.Expression(
    r"input\1.pcn")
print(f)
```

```python
ordering = list(range(1, f.numVars + 1))
for i in range(len(ordering) + 1, -1, -1):
    print(i)
    a = BDD(f, ordering, i)
    a.displayGraph()
```

```python
import itertools

ordering = list(range(1, f.numVars + 1))[::-1]
best = len(ordering) ** len(ordering)
for p in itertools.permutations(ordering):
    print(p)
    a = BDD(f, p)
    node_cnt = a.getNodeCount()
    print(node_cnt)
    if node_cnt < best:
        best = node_cnt
        a.displayGraph()
```

```python
f = boolfunc.Expression(r"input\1.pcn")
g = boolfunc.Expression(r"input\3.pcn")
print(f)
print()
print(g)
print()
print(f.andExp(g.cubes))
```

```python
ordering = list(range(1, max(f.numVars, g.numVars) + 1))

bdd_f = BDD(f, ordering)
bdd_g = BDD(g, ordering)

cache = bdd_f.NODES
for n in bdd_g.NODES:
    cache[n] = bdd_g.NODES[n]

print(bdd_f.exp)
op = applyAnd(bdd_f.node, bdd_g.node, ordering, cache)
op.displayGraph()
```
